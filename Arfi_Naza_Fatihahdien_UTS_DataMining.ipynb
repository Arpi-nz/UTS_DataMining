SOAL 1

```python
temperatures = [25, 28, 30, 22, 35, 40, 18, 24]

# 1. Menggunakan list comprehension untuk konversi ke Fahrenheit
temp_fahrenheit = [(c * 9/5) + 32 for c in temperatures]

# 2. Mengambil 3 data suhu terakhir menggunakan slicing
tiga_suhu_terakhir = temp_fahrenheit[-3:]

# Menampilkan hasil untuk memverifikasi jawaban
print("Suhu dalam Fahrenheit:", temp_fahrenheit)
print("3 data suhu terakhir:", tiga_suhu_terakhir)
```

SOAL 2

```python
inventory = {
    "laptop": {"stok": 10, "harga": 15000000},
    "mouse": {"stok": 50, "harga": 250000},
    "monitor": {"stok": 20, "harga": 3000000}
}

# 1. Menambahkan produk baru: "keyboard"
inventory["keyboard"] = {"stok": 30, "harga": 500000}

# 2. Memperbarui harga "laptop" menggunakan metode .update()
inventory["laptop"].update({"harga": 14500000})

# 3. Menghitung total nilai aset di gudang (stok dikali harga untuk semua barang)
total_aset = sum(barang["stok"] * barang["harga"] for barang in inventory.values())

# Menampilkan hasil untuk memverifikasi jawaban
print("Inventory saat ini:", inventory)
print("Total nilai aset di gudang: Rp", total_aset)
```

```python
import numpy as np

scores = np.array([
    [80, 85, 90],
    [70, 60, 75],
    [95, 90, 100],
    [40, 50, 45]
])

# 1. Menghitung rata-rata nilai untuk setiap mahasiswa (axis=1)
rata_rata_mahasiswa = np.mean(scores, axis=1)

# 2. Mencari nilai tertinggi dari seluruh data
nilai_tertinggi = np.max(scores)

# 3. Menggunakan np.where untuk membuat array status kelulusan
status_kelulusan = np.where(rata_rata_mahasiswa >= 70, "Lulus", "Gagal")

# Menampilkan hasil untuk memverifikasi jawaban
print("Rata-rata setiap mahasiswa:", rata_rata_mahasiswa)
print("Nilai tertinggi dari seluruh data:", nilai_tertinggi)
print("Status Kelulusan:", status_kelulusan)
```

```python
import pandas as pd
import numpy as np

df_data = {
    'Nama': ['Andi', 'Budi', 'Caca', 'Dedi', 'Euis'],
    'Departemen': ['IT', 'HR', 'IT', 'Sales', 'Sales'],
    'Gaji': [8000000, 7000000, 8500000, np.nan, 6000000],
    'Pengalaman_Tahun': [5, 3, 6, 2, 1]
}

df = pd.DataFrame(df_data)

# 1. Mengisi nilai Gaji yang kosong (NaN) dengan median dari seluruh gaji
median_gaji = df['Gaji'].median()
df['Gaji'] = df['Gaji'].fillna(median_gaji)

# 2. Membuat kolom baru 'Kategori_Senioritas' menggunakan np.where
df['Kategori_Senioritas'] = np.where(df['Pengalaman_Tahun'] > 4, 'Senior', 'Junior')

# (Catatan: Jika ingin menggunakan .apply(), kodenya seperti ini:
# df['Kategori_Senioritas'] = df['Pengalaman_Tahun'].apply(lambda x: 'Senior' if x > 4 else 'Junior') )

# 3. Menampilkan rata-rata Gaji per Departemen menggunakan .groupby()
rata_gaji_per_dept = df.groupby('Departemen')['Gaji'].mean()

# Menampilkan hasil
print("DataFrame setelah diperbarui:")
print(df)
print("\nRata-rata Gaji per Departemen:")
print(rata_gaji_per_dept)
```
